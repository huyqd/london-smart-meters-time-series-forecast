# Introduction to Target Transformation in Machine Learning Time Series Forecasting

When building machine learning models for time series forecasting, one of the most important (and often overlooked) steps is **target transformation**—the process of modifying the target variable ($y_t$) before modeling. This step can dramatically improve model performance, stability, and interpretability, especially when dealing with real-world data that often exhibits trends, seasonality, or non-constant variance.

---

## Why Transform the Target Variable in ML Forecasting?

Let's break down the reasons in a beginner-friendly way:

- **Stabilize Variance:**  
    Many time series, like electricity demand, show increasing or decreasing variability over time. This means the "ups and downs" get bigger or smaller as time goes on. Transformations like the logarithm or Box-Cox can help make the ups and downs more consistent, which is easier for machine learning models to handle.

- **Remove Trends and Seasonality:**  
    Time series data often has patterns that repeat (seasonality) or steadily increase/decrease (trend). For example, electricity usage might be higher in the evenings or on weekends. By "differencing" (subtracting previous values), we can remove these predictable patterns, allowing the model to focus on what’s truly unpredictable.

- **Handle Non-Normality:**  
    Many machine learning algorithms work best when the data is roughly "bell-shaped" (normally distributed). If the data is skewed or has outliers, transformations can help make it more suitable for these models.

- **Improve Forecast Accuracy:**  
    By making the series more "stationary" (its statistical properties don’t change over time), transformations often lead to more accurate and reliable forecasts.

---

## Common Target Transformations for ML Time Series

| Transformation      | Purpose                                  | Formula                                 |
|---------------------|------------------------------------------|-----------------------------------------|
| **Logarithm**       | Stabilize variance, handle skewness      | $y'_t = \log(y_t)$                      |
| **Box-Cox**         | Generalized variance stabilization       | $y'_t = \frac{y_t^\lambda - 1}{\lambda}$ (for $\lambda \neq 0$) |
| **Differencing**    | Remove trend/seasonality                 | $y'_t = y_t - y_{t-k}$                  |
| **Standardization** | Center and scale                         | $y'_t = \frac{y_t - \mu}{\sigma}$       |

**Beginner Tip:**  
- *Logarithm* is great when your data has values that grow exponentially or have large spikes.
- *Differencing* is your go-to for removing repeating patterns or steady growth/decline.
- *Standardization* helps when you want your data to have a mean of 0 and a standard deviation of 1, which is often required for neural networks and other ML models.

---

## Real-World Analogy

Imagine you're trying to predict how much electricity a city will use tomorrow. If the city's population is growing, the total usage will naturally trend upward over time. If you don't adjust for this trend, your model might just "learn" that usage always goes up, missing the real factors that cause day-to-day changes (like weather or holidays). Target transformation helps your model focus on the meaningful signals, not just the obvious trends.

## Should We Deseasonalize the Data or Model Seasonality Directly?

This is a classic question in time series forecasting, especially for beginners!

### What is Seasonality?

**Seasonality** refers to patterns that repeat at regular intervals in your data. For example, electricity usage might spike every evening or drop every weekend. In mathematical terms, a seasonal component can be represented as:

$$
y_t = \text{Trend}_t + \text{Seasonality}_t + \text{Noise}_t
$$

### Two Main Approaches

#### 1. **Deseasonalize the Data (Remove Seasonality Before Modeling)**

- **How:**  
    You can "deseasonalize" by applying transformations like seasonal differencing (subtracting the value from the same period in the previous cycle, e.g., $y_t - y_{t-48}$ for daily seasonality in half-hourly data).
- **Why:**  
    This makes the data more stationary, which many machine learning models prefer. The model then only needs to learn the unpredictable part (the "noise").
- **Downside:**  
    After forecasting, you must "reseasonalize" (add the seasonal component back) to interpret predictions in the original scale.

#### 2. **Model the Seasonality Directly**

- **How:**  
    Instead of removing seasonality, you can provide the model with features that help it learn seasonal patterns. For example, you might add:
    - **Time-of-day** or **day-of-week** indicators
    - **Fourier terms** (sine/cosine functions to capture cycles)
- **Why:**  
    Many modern ML models (like tree-based models or neural networks) can learn complex seasonal patterns if you give them the right features.
- **Downside:**  
    If the seasonality is very strong and regular, not removing it can make the model's job harder, especially for simpler models.

---

### Which Approach Should You Use?

- **For classical statistical models** (like ARIMA), deseasonalizing is often essential.
- **For machine learning models**, it's common to **model seasonality directly** by engineering features, but you can also combine both approaches for best results.

**Beginner Tip:**  
If you're unsure, start by modeling seasonality with features. If your model struggles, try deseasonalizing and see if performance improves!

## Should We Detrend the Data or Model the Trend Directly?

Just like with seasonality, **trend** is a fundamental component of many time series. Let's break down what trend means, why it matters, and the two main approaches for handling it in machine learning forecasting.

---

### What is Trend?

**Trend** refers to the long-term increase or decrease in your data over time. For example, electricity consumption in a growing city might steadily rise year after year, while the popularity of a fading technology might show a downward trend.

Mathematically, a time series with trend can be represented as:

$$
y_t = \text{Trend}_t + \text{Seasonality}_t + \text{Noise}_t
$$

- **Trend$_t$**: The underlying direction (upward or downward) over time.
- **Seasonality$_t$**: Repeating patterns at regular intervals.
- **Noise$_t$**: Random fluctuations.

---

### Two Main Approaches

#### 1. **Detrend the Data (Remove Trend Before Modeling)**

- **How:**  
    You can "detrend" by subtracting the trend component from your data. The simplest method is *first differencing*:  
    $$
    y'_t = y_t - y_{t-1}
    $$
    This transformation removes the linear trend, making the series more stationary (its properties don't change over time).

- **Why:**  
    Many machine learning models work better with stationary data. By removing the trend, you help the model focus on the unpredictable parts of the series.

- **Downside:**  
    After forecasting, you must "retrend" (add the trend back) to interpret predictions in the original scale. This can be tricky if the trend is complex or changes over time.

#### 2. **Model the Trend Directly**

- **How:**  
    Instead of removing the trend, you can provide the model with features that help it learn the trend. Common features include:
    - **Time index** (e.g., a column counting the number of periods since the start)
    - **Polynomial terms** (e.g., squared or cubic time index for nonlinear trends)
    - **Rolling averages** (to smooth out short-term fluctuations)

- **Why:**  
    Modern ML models (like tree-based models or neural networks) can learn complex trends if you give them the right features. This approach keeps your predictions in the original scale, making interpretation easier.

- **Downside:**  
    If the trend is very strong or non-linear, the model might need more sophisticated features or architectures to capture it accurately.

---

### Which Approach Should You Use?

- **For classical statistical models** (like ARIMA), detrending is often essential.
- **For machine learning models**, it's common to **model the trend directly** by engineering features, but you can also combine both approaches for best results.

**Beginner Tip:**  
If you're just starting out, try modeling the trend with features first. If your model struggles, experiment with detrending and see if performance improves!

## Understanding the Order of Transformations in MLForecast

When working with time series forecasting using the `MLForecast` library, it's important to understand **the order in which data transformations are applied**. This order can affect both your model’s features and its predictions.

### The Transformation Pipeline

**MLForecast** applies transformations in the following sequence:

1. **Target Transformation**  
    The target variable ($y_t$) is transformed first. This could be standardization, differencing, log transformation, etc.

2. **Lag and Lag-Based Feature Creation**  
    After the target is transformed, lag features (like $y_{t-1}$, $y_{t-48}$, or rolling means) are computed **using the transformed target**.

#### Why Does This Matter?

- If you apply a transformation like standardization or differencing, the lag features will be based on the **transformed** values, not the original ones.
- For example, if you difference the target ($y'_t = y_t - y_{t-1}$), then the lag-1 feature will be $y'_{t-1} = y_{t-1} - y_{t-2}$, not just the original $y_{t-1}$.

### Visualizing the Process

Suppose you have a time series:

$$
y_1, \ y_2, \ y_3, \ \ldots, \ y_t
$$

If you apply a transformation $T$ (like standardization or differencing), you get:

$$
y'_t = T(y_t)
$$

Then, lag features are created from $y'_t$:

- **Lag 1:** $y'_{t-1}$
- **Lag 2:** $y'_{t-2}$
- **Rolling Mean (window=3):** $\frac{y'_{t-1} + y'_{t-2} + y'_{t-3}}{3}$

### Key Takeaway

> **In MLForecast, target transformations are applied before lag features are created.**  
> This means all lag-based features are computed from the transformed target, not the original series.

**Beginner Tip:**  
Always keep this order in mind when interpreting your features and results. If you want lag features from the original series, you’ll need to create them manually before applying target transformations.
